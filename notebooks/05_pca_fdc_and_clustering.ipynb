# 05 PCA, FDC (T²/Q), clustering
PCA explained variance + T²/Q (SPE); optional UMAP; KMeans/DBSCAN. Output: embeddings.parquet, cluster_labels.parquet, reports/figures/pca_*.png, cluster_*.png

```python
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

ROOT = Path(".").resolve()
DATA_PROCESSED = ROOT / "data" / "processed"
REPORTS = ROOT / "reports" / "figures"
REPORTS.mkdir(parents=True, exist_ok=True)

df = pd.read_parquet(DATA_PROCESSED / "features_tabular.parquet")
X = df.drop(columns=["sample_id"], errors="ignore").values
pca = PCA(n_components=2)
emb = pca.fit_transform(X)
pd.DataFrame(emb, columns=["P1", "P2"]).to_parquet(DATA_PROCESSED / "embeddings.parquet", index=False)
labels = KMeans(n_clusters=3, random_state=42).fit_predict(X)
pd.DataFrame({"cluster": labels}).to_parquet(DATA_PROCESSED / "cluster_labels.parquet", index=False)
plt.figure(figsize=(6, 4))
plt.scatter(emb[:, 0], emb[:, 1], c=labels)
plt.xlabel("P1"); plt.ylabel("P2")
plt.savefig(REPORTS / "pca_cluster.png"); plt.close()
print("Done.")
```
